```python
import numpy as np
temp_max = np.array([17, 19, 21, 28, 33, 38, 37, 37, 31, 23, 19, 18])
temp_min = np.array([-62, -59, -56, -46, -32, -18, -9, -13, -25, -46, -52, -58])
import matplotlib.pyplot as plt
months = np.arange(12)
plt.plot(months, temp_max, "ro")
plt.plot(months, temp_min, "bo")
plt.xlabel("Month")
plt.ylabel("Min and max temperature")

coeff1=np.polyfit(months,temp_max, deg=3)
poly1=np.poly1d(coeff1)

y_pred1=poly1(months)
mae1=np.mean(np.abs(temp_max-y_pred1))
rmse=np.sqrt(np.mean((temp_max-y_pred1)**2))


coeff2=np.polyfit(months,temp_min,deg=3)
poly2=np.poly1d(coeff2)

y_pred2=poly2(months)
mae2=np.mean(np.abs(temp_min-y_pred2))
rmse2=np.sqrt(np.mean((temp_min-y_pred2)**2))
plt.figure()
plt.plot(months,temp_max, label="Dati temp massimi normali")
plt.plot(months,y_pred1,label=f"Date temp massimi fittati:{mae1:.2f},{rmse:.2f}")

plt.plot(months,temp_min,label="Dati temp min normali")
plt.plot(months,y_pred2, label=f"Date temp minimi fittati:{mae2:.2f},{rmse2:.2f}")
plt.legend()
plt.show()

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error

# Carica il dataset
df = pd.read_csv('https://gist.githubusercontent.com/seankross/a412dfbd88b3db70b74b/raw/5f23f993cd87c283ce766e7ac6b329ee7cc2e1d1/mtcars.csv')

# Seleziona le colonne
X = df[['disp']].values  # Feature deve essere 2D
y = df['mpg'].values     # Target

# Crea e addestra il modello
model = LinearRegression()
model.fit(X, y)

# Predizioni
y_pred = model.predict(X)

# Coefficienti
slope = model.coef_[0]
intercept = model.intercept_
print(f"Equazione della retta: mpg = {slope:.4f} * disp + {intercept:.4f}")

# Calcolo errori
mae = mean_absolute_error(y, y_pred)
rmse = mean_squared_error(y, y_pred, squared=False)
print(f"MAE: {mae:.2f}")
print(f"RMSE: {rmse:.2f}")

# Plot
plt.scatter(X, y, color='blue', label='Dati reali')
plt.plot(X, y_pred, color='red', label='Regressione lineare')
plt.xlabel('Displacement (disp)')
plt.ylabel('Miles per Gallon (mpg)')
plt.title('Regressione Lineare: mpg vs disp')
plt.legend()
plt.show()
```
