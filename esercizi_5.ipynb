```python
# Import dei pacchetti
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Caricamento del dataset
url = 'https://zenodo.org/record/5898311/files/vgsales.csv'
df = pd.read_csv(url)
df.head()

pop=df.groupby('Genre').sum()

pop.sort_values(by='Rank',ascending=False,inplace=True)

pop=pop.head(5)





pop=pop[['NA_Sales','EU_Sales','JP_Sales','Other_Sales']]

pop.plot(kind="bar")




```
